```python
import pandas as pd
import os

# Load file
try:
    df = pd.read_csv('Updated_sales.csv')
except UnicodeDecodeError:
    df = pd.read_csv('Updated_sales.csv', encoding='latin-1')

df.columns = df.columns.astype(str).str.strip()

def find_col(keywords, exclude=()):
    for kw in keywords:
        for c in df.columns:
            n = c.lower().replace('_', ' ')
            if kw in n and not any(e in n for e in exclude):
                return c
    return None

DATE     = find_col(['date', 'time'])
PRICE    = find_col(['unit price', 'price', 'mrp', 'rate'])
QTY      = find_col(['quantity', 'qty', 'units'])
CATEGORY = find_col(['category', 'segment', 'type'])
BRAND    = find_col(['brand', 'company', 'manufacturer'])
REGION   = find_col(['region', 'state', 'zone', 'area', 'city', 'country'])
PRODUCT  = find_col(['product', 'item', 'name'], exclude=['brand', 'category'])

print("Columns in file:", df.columns.tolist(), "\n")
print("Detected:")
for k, v in [('DATE', DATE), ('PRODUCT', PRODUCT), ('CATEGORY', CATEGORY),
             ('BRAND', BRAND), ('REGION', REGION), ('PRICE', PRICE), ('QTY', QTY)]:
    print(f"  {k:9} -> {v}")
df.head()
```

```python
# Revenue
if PRICE and QTY:
    df[PRICE] = pd.to_numeric(df[PRICE], errors='coerce')
    df[QTY] = pd.to_numeric(df[QTY], errors='coerce')
    df['Revenue'] = df[PRICE] * df[QTY]
else:
    rev_col = find_col(['revenue', 'sales', 'amount', 'total'])
    df['Revenue'] = pd.to_numeric(df[rev_col], errors='coerce')

# Date parts
if DATE:
    parsed = pd.to_datetime(df[DATE], errors='coerce')
    if parsed.notna().mean() < 0.5:
        parsed = pd.to_datetime(df[DATE], errors='coerce', dayfirst=True)
    df[DATE] = parsed
    df['Year'] = df[DATE].dt.year
    df['Month'] = df[DATE].dt.month
    df['Month_Name'] = df[DATE].dt.month_name()
    df['Year_Month'] = df[DATE].dt.to_period('M').astype(str)

df.head()
```

```python
threshold = df['Revenue'].median()          # sales above this value
high_sales = df[df['Revenue'] > threshold]
print(f"Rows with Revenue > {threshold:,.2f}: {len(high_sales)}")

if CATEGORY:
    top_category = df[CATEGORY].value_counts().index[0]   # most common category
    cat_df = df[df[CATEGORY] == top_category]
    combo = df[(df[CATEGORY] == top_category) & (df['Revenue'] > threshold)]
    print(f"Category '{top_category}': {len(cat_df)} rows | both filters: {len(combo)} rows")
    display(cat_df.head())
```

```python
def rev_by(col):
    return (df.groupby(col, as_index=False)['Revenue'].sum()
              .sort_values('Revenue', ascending=False))

rev_by_category = rev_by(CATEGORY) if CATEGORY else None
rev_by_brand    = rev_by(BRAND) if BRAND else None
rev_by_region   = rev_by(REGION) if REGION else None
rev_by_month    = (df.groupby('Year_Month', as_index=False)['Revenue'].sum()
                     .sort_values('Year_Month')) if DATE else None

for name, d in [('Revenue by category', rev_by_category), ('Revenue by brand', rev_by_brand),
                ('Revenue by region', rev_by_region), ('Revenue by month', rev_by_month)]:
    if d is not None:
        print(name)
        display(d.head(10))
```

```python
if PRODUCT:
    product_rev = df.groupby(PRODUCT, as_index=False)['Revenue'].sum()
    product_rev['Rank'] = product_rev['Revenue'].rank(ascending=False, method='dense').astype(int)
    top10_products = product_rev.nlargest(10, 'Revenue')
else:
    top10_products = df.nlargest(10, 'Revenue')

top10_products
```

```python
pivot_cat_region = None
if CATEGORY and REGION:
    pivot_cat_region = pd.pivot_table(
        df, values='Revenue', index=CATEGORY, columns=REGION,
        aggfunc='sum', fill_value=0, margins=True, margins_name='Total'
    )
    display(pivot_cat_region)
else:
    print("Category or Region column not found, pivot skipped.")
```

```python
summary_stats = pd.DataFrame({
    'Metric': ['Total Revenue', 'Mean Revenue', 'Median Revenue', 'Std Dev', 'Max', 'Min', 'Rows'],
    'Value': [df['Revenue'].sum(), df['Revenue'].mean(), df['Revenue'].median(),
              df['Revenue'].std(), df['Revenue'].max(), df['Revenue'].min(), len(df)]
})
if QTY:
    summary_stats.loc[len(summary_stats)] = ['Total Units', df[QTY].sum()]

summary_stats
```

```python
os.makedirs('summaries', exist_ok=True)

summaries = {
    'rev_by_category': rev_by_category,
    'rev_by_brand': rev_by_brand,
    'rev_by_region': rev_by_region,
    'rev_by_month': rev_by_month,
    'top10_products': top10_products,
    'pivot_cat_region': pivot_cat_region.reset_index() if pivot_cat_region is not None else None,
    'summary_stats': summary_stats,
}
summaries = {k: v for k, v in summaries.items() if v is not None}

for name, d in summaries.items():
    d.to_csv(f'summaries/{name}.csv', index=False)

with pd.ExcelWriter('summaries/summary_tables.xlsx') as writer:
    for name, d in summaries.items():
        d.to_excel(writer, sheet_name=name[:31], index=False)

print("Saved:", list(summaries.keys()))

from google.colab import files
files.download('summaries/summary_tables.xlsx')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# remove blank rows and duplicates
data = df.dropna(subset=['Revenue', 'Order Date'])
data = data.drop_duplicates()
data = data[data['Order Date'].dt.year > 2000].copy()

data['City'] = data['Purchase Address'].str.split(',').str[1].str.strip()
data['Hour'] = data['Order Date'].dt.hour
data['Month'] = data['Order Date'].dt.strftime('%Y-%m')

monthly = data.groupby('Month')['Revenue'].sum()
hourly = data['Hour'].value_counts().sort_index()
by_city = data.groupby('City')['Revenue'].sum().sort_values(ascending=False)
prod = data.groupby('Product').agg(Units=('Quantity Ordered', 'sum'),
                                   Revenue=('Revenue', 'sum'),
                                   Price=('Price Each', 'mean'))
```

```python
# Line Chart
plt.figure(figsize=(8, 5))
plt.plot(hourly.index, hourly.values, color='red', marker='o', linestyle='-', label='Orders')
plt.title('Orders by Hour')
plt.xlabel('Hour')
plt.ylabel('Number of orders')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Vertical bar graph
plt.figure(figsize=(8, 5))
plt.bar(monthly.index, monthly.values, color='teal')
plt.title('Revenue by Month')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(axis='y')
plt.show()
```

```python
# horizontal bargraph
plt.figure(figsize=(8, 5))
plt.barh(by_city.index, by_city.values, color='orange')
plt.title('Revenue by City')
plt.xlabel('Revenue')
plt.ylabel('City')
plt.grid(axis='x')
plt.show()
```

```python
# pie chart
top = prod['Revenue'].sort_values(ascending=False).head(6)

plt.figure(figsize=(7, 7))
plt.pie(top.values, labels=top.index, autopct='%1.1f%%', startangle=90, explode=[0.1, 0, 0, 0, 0, 0])
plt.title('Top 6 Products by Revenue')
plt.show()
```

```python
# Histogram
plt.figure(figsize=(8, 5))
plt.hist(data['Price Each'], bins=20, color='skyblue', edgecolor='black')
plt.title('Price Distribution')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```

```python
# Scatterplot with Colorbar
plt.figure(figsize=(8, 6))
sc = plt.scatter(prod['Price'], prod['Units'], c=prod['Revenue'], s=80, cmap='viridis')
plt.colorbar(sc, label='Revenue')
plt.xscale('log')
plt.title('Price vs Units Sold')
plt.xlabel('Average price')
plt.ylabel('Units sold')
plt.grid(True)
plt.show()
```

```python
soup=BeautifulSoup(response.text, 'html.parser')
print(soup)
```

```python
table = soup.find('table')
print(table)

if table is None:
  print('Table not found')
else:
  print('Table found  fetched')
```

```python
rows = table.find_all("tr")

print("Number of rows:", len(rows))
```

```python
data=[]
for row in rows:
    # Find all header and data cells in the current row
    cells = row.find_all(["th", "td"])

    # Initialize row_data as a list
    row_data = []

    for cell in cells:
        value = cell.get_text(strip=True)
        row_data.append(value)

    data.append(row_data)

print(data)
```
